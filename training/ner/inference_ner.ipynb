```python
from pathlib import Path
from transformers import pipeline


def load_ner_pipeline(model_path: Path):
    print(f"Loading NER model from: {model_path}")
    return pipeline(
        task="token-classification",
        model=str(model_path),
        tokenizer=str(model_path),
        aggregation_strategy="simple",
    )


def run_ner(ner_pipeline, text: str):
    print("Starting NER inference...")
    results = ner_pipeline(text)
    print("NER inference completed.")
    return results


def main():
    model_dir = Path("training/ner/done/hf/ner_model").resolve()
    text = "The invention improves the mould container parts filled with green bricks."

    print("Initializing NER pipeline...")
    ner = load_ner_pipeline(model_dir)

    print("Input text:")
    print(text)

    entities = run_ner(ner, text)

    print("Extracted entities:")
    for entity in entities:
        print(entity)

    print("Process finished successfully.")


if __name__ == "__main__":
    main()
```
